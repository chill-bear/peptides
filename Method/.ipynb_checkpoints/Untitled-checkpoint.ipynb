```python

import pandas as pd
from Bio.SeqUtils.ProtParam import ProteinAnalysis

data = {
    "Peptide": [
        "FKKIEKAVRRVR",  # active
        "GRMRWKWIKKRI",
        "AAAAAAAAGGGG",  # inactive
        "VVVVVVVVVVVV"
    ],
    "Group": [
        "Top Active",
        "Top Active",
        "Bottom Inactive",
        "Bottom Inactive"
    ]
}

df = pd.DataFrame(data)


def compute_net_charge(seq, pH=7.0):
    return ProteinAnalysis(seq).charge_at_pH(pH)

def compute_gravy(seq):
    return ProteinAnalysis(seq).gravy()

def aa_ratio(seq, aa_list):
    return sum(seq.count(aa) for aa in aa_list) / len(seq)

df["NetCharge_pH7"] = df["Peptide"].apply(compute_net_charge)
df["GRAVY"] = df["Peptide"].apply(compute_gravy)
df["PosChargeRatio"] = df["Peptide"].apply(lambda s: aa_ratio(s, "KR"))
df["AromaticRatio"] = df["Peptide"].apply(lambda s: aa_ratio(s, "FWY"))


display(df)
```
