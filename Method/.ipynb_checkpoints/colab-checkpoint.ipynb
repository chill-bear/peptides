```python
import torch
import torch.nn as nn
import torch.optim as optim
from Bio import SeqIO
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# ==== Step 1: Amino acid vocab + encoding ====
amino_acids = list("ACDEFGHIKLMNPQRSTVWY")
aa_to_idx = {aa: i+1 for i, aa in enumerate(amino_acids)}  # 0 is reserved for padding
vocab_size = len(aa_to_idx) + 1  # includes padding

def encode_sequence(seq, maxlen):
    idx_seq = [aa_to_idx.get(aa, 0) for aa in seq]
    return idx_seq + [0] * (maxlen - len(idx_seq))

# ==== Step 2: Load model1.csv for regression ====
df_model1 = pd.read_csv("/Users/jiayingyou/Desktop/model1.csv")
seqs_reg = df_model1["Sequence"].tolist()
labels_reg = df_model1["IC50 Biofilm"].values

# ==== Step 3: Load FASTA for classification ====
positive = list(SeqIO.parse("/Users/jiayingyou/Documents/positive.fasta", "fasta"))
negative = list(SeqIO.parse("/Users/jiayingyou/Documents/negative.fasta", "fasta"))
seqs_cls = [str(r.seq) for r in positive + negative]
labels_cls = [1]*len(positive) + [0]*len(negative)

# ==== Step 4: Pad + encode ====
maxlen = max(max(len(s) for s in seqs_reg), max(len(s) for s in seqs_cls))
X1 = torch.tensor([encode_sequence(s, maxlen) for s in seqs_reg], dtype=torch.long)
y1 = torch.tensor(labels_reg, dtype=torch.float32).view(-1, 1)

X2 = torch.tensor([encode_sequence(s, maxlen) for s in seqs_cls], dtype=torch.long)
y2 = torch.tensor(labels_cls, dtype=torch.long)

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=42)

# ==== Step 5: BiLSTM Model ====
class BiLSTMModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=16, hidden_dim=8, output_dim=1, task='regression'):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True, bidirectional=True)
        self.head = nn.Sequential(
            nn.Linear(hidden_dim*2, 8),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(8, output_dim)
        )
        self.task = task

    def forward(self, x):
        x = self.embedding(x)                          # (batch, seq_len, embed)
        _, (h_n, _) = self.lstm(x)                     # h_n: (2, batch, hidden)
        h = torch.cat((h_n[0], h_n[1]), dim=1)         # (batch, hidden*2)
        out = self.head(h)
        return out

# ==== Step 6: Train Model 1 (Regression) ====
model1 = BiLSTMModel(vocab_size, task='regression')
loss_fn1 = nn.MSELoss()
optimizer1 = optim.Adam(model1.parameters(), lr=0.001)

for epoch in range(300):
    model1.train()
    optimizer1.zero_grad()
    pred = model1(X1_train)
    loss = loss_fn1(pred, y1_train)
    loss.backward()
    optimizer1.step()
    if epoch % 10 == 0:
        print(f"[Reg] Epoch {epoch}, Loss: {loss.item():.4f}")

# ==== Step 7: Save Model 1 weights ====
torch.save(model1.state_dict(), "bilstm_model1_regression.pt")

# ==== Step 8: Train Model 2 (Classification) with shared BiLSTM ====
model2 = BiLSTMModel(vocab_size, output_dim=2, task='classification')
model2.embedding = model1.embedding
model2.lstm = model1.lstm
for param in model2.embedding.parameters(): param.requires_grad = False
for param in model2.lstm.parameters(): param.requires_grad = False

loss_fn2 = nn.CrossEntropyLoss()
optimizer2 = optim.Adam(filter(lambda p: p.requires_grad, model2.parameters()), lr=0.001)

for epoch in range(200):
    model2.train()
    optimizer2.zero_grad()
    logits = model2(X2_train)
    loss = loss_fn2(logits, y2_train)
    loss.backward()
    optimizer2.step()
    if epoch % 10 == 0:
        print(f"[CLS] Epoch {epoch}, Loss: {loss.item():.4f}")

# ==== Step 9: Evaluate Classification ====
model2.eval()
with torch.no_grad():
    pred_cls = model2(X2_test).argmax(dim=1)
    acc = (pred_cls == y2_test).float().mean().item()
print(f"\n✅ Final Classification Accuracy: {acc:.2f}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from Bio import SeqIO
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from transformers import AutoModel, AutoTokenizer
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

# ==== Step 1: Setup ==== 
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ==== Step 2: Load ESM2 model (8M) ====
model_name = "facebook/esm2_t6_8M_UR50D"
esm_model = AutoModel.from_pretrained(model_name).to(device)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# ==== Step 3: Define ESM embedding extraction ====
def extract_esm_embeddings(seqs, maxlen=512, batch_size=16):
    embeddings = []
    esm_model.eval()
    with torch.no_grad():
        for i in tqdm(range(0, len(seqs), batch_size)):
            batch = seqs[i:i+batch_size]
            tokens = tokenizer(batch, return_tensors="pt", padding=True, truncation=True,
                               max_length=maxlen, add_special_tokens=True).to(device)
            out = esm_model(**tokens)
            reps = out.last_hidden_state
            masks = tokens['attention_mask']
            summed = (reps * masks.unsqueeze(-1)).sum(1)
            lens = masks.sum(1).unsqueeze(-1)
            mean = summed / lens
            embeddings.append(mean.cpu())
    return torch.cat(embeddings, dim=0).numpy()

# ==== Step 4: Load sequences and labels ====
df_model1 = pd.read_csv("/Users/jiayingyou/Desktop/model1.csv")
seqs_reg = df_model1["Sequence"].tolist()
y1 = torch.tensor(df_model1["IC50 Biofilm"].values, dtype=torch.float32).view(-1, 1)

pos = list(SeqIO.parse("/Users/jiayingyou/Documents/positive.fasta", "fasta"))
neg = list(SeqIO.parse("/Users/jiayingyou/Documents/negative.fasta", "fasta"))
seqs_cls = [str(r.seq) for r in pos + neg]
y2 = torch.tensor([1]*len(pos) + [0]*len(neg), dtype=torch.long)

# ==== Step 5: Extract ESM embeddings ====
print("Extracting ESM embeddings for regression data...")
X1 = extract_esm_embeddings(seqs_reg)
print("Extracting ESM embeddings for classification data...")
X2 = extract_esm_embeddings(seqs_cls)

# ==== Step 6: Prepare train/test splits ====
X1_train, X1_test, y1_train, y1_test = train_test_split(torch.tensor(X1), y1, test_size=0.2, random_state=42)
X2_train, X2_test, y2_train, y2_test = train_test_split(torch.tensor(X2), y2, test_size=0.2, random_state=42)

# ==== Step 7: Define BiLSTM-like downstream head ====
class FeedForward(nn.Module):
    def __init__(self, input_dim=320):
        super().__init__()
        self.base = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(64, 8),
            nn.ReLU(),
        )

    def forward(self, x):
        return self.base(x)

# ==== Step 8: Regression model ====
class Model1_Reg(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = FeedForward()
        self.out = nn.Linear(8, 1)

    def forward(self, x):
        x = self.encoder(x)
        return self.out(x)

model1 = Model1_Reg().to(device)
loss_fn1 = nn.MSELoss()
optimizer1 = optim.Adam(model1.parameters(), lr=1e-3)

best_loss = float('inf')
patience, counter = 30, 0

print("Training regression model...")
for epoch in range(150):
    model1.train()
    optimizer1.zero_grad()
    pred = model1(X1_train.to(device))
    loss = loss_fn1(pred, y1_train.to(device))
    loss.backward()
    optimizer1.step()
    
    model1.eval()
    with torch.no_grad():
        val_loss = loss_fn1(model1(X1_test.to(device)), y1_test.to(device)).item()
    
    if val_loss < best_loss:
        best_loss = val_loss
        torch.save(model1.state_dict(), "esm_model1_reg.pt")
        counter = 0
    else:
        counter += 1
        if counter >= patience:
            print(f"Early stopping at epoch {epoch}")
            break
    
    if epoch % 10 == 0:
        print(f"[Reg Epoch {epoch}] Train Loss: {loss.item():.4f}, Val Loss: {val_loss:.4f}")

# ==== Step 9: Classification model ====
class Model2_Cls(nn.Module):
    def __init__(self, pretrained_encoder):
        super().__init__()
        self.encoder = pretrained_encoder.encoder
        for param in self.encoder.parameters():
            param.requires_grad = False
        self.out = nn.Sequential(
            nn.Linear(8, 4),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(4, 2)
        )

    def forward(self, x):
        x = self.encoder(x)
        return self.out(x)

model1_loaded = Model1_Reg()
model1_loaded.load_state_dict(torch.load("esm_model1_reg.pt"))
model2 = Model2_Cls(model1_loaded).to(device)

loss_fn2 = nn.CrossEntropyLoss()
optimizer2 = optim.Adam(filter(lambda p: p.requires_grad, model2.parameters()), lr=1e-3)

best_acc, counter = 0.0, 0

print("Training classification model...")
for epoch in range(150):
    model2.train()
    optimizer2.zero_grad()
    logits = model2(X2_train.to(device))
    loss = loss_fn2(logits, y2_train.to(device))
    loss.backward()
    optimizer2.step()

    model2.eval()
    with torch.no_grad():
        preds = model2(X2_test.to(device)).argmax(dim=1)
        acc = (preds == y2_test.to(device)).float().mean().item()
    
    if acc > best_acc:
        best_acc = acc
        torch.save(model2.state_dict(), "esm_model2_cls.pt")
        counter = 0
    else:
        counter += 1
        if counter >= patience:
            print(f"Early stopping at epoch {epoch}")
            break

    if epoch % 10 == 0:
        print(f"[Cls Epoch {epoch}] Loss: {loss.item():.4f}, Acc: {acc:.4f}")

print(f"\n✅ Final Test Accuracy (Model 2): {best_acc:.2f}")
```
